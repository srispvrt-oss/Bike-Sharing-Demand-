```python
# Function to apply IQR method to fix outliers
def fix_outliers_iqr(df):
  # only the measured weather columns are capped; codes (season, hr, ...) and counts are not touched
  num = df.select_dtypes(include='number').columns.drop(
      ['cnt','casual','registered','season','yr','mnth','hr',
        'holiday','weekday','workingday','weathersit']
  )
  for col in num:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        # Replace outliers with the respective bounds
        df[col] = df[col].clip(
            lower=lower_bound,
            upper=upper_bound
        )
  return df


```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.feature_selection import SelectKBest, f_regression

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.neighbors import KNeighborsRegressor

from sklearn.metrics import (
     mean_absolute_error,
     mean_squared_error,
     r2_score
)
RANDOM_STATE = 42
sns.set_theme(style="whitegrid")
```

```python
# Select columns for IQR capping
boxplot_columns = [
   column for column in ["temp", "atemp", "hum", "windspeed"]
   if column in df.columns
]
# Apply IQR Capping
for column in boxplot_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df[column] = df[column].clip(
        lower=lower_bound,
        upper=upper_bound
    )
# Set the plotting style
sns.set_theme(style="whitegrid")
# Plot boxplots after IQR capping
plt.figure(figsize=(12, 5))
sns.boxplot(data=df[boxplot_columns])
plt.title("Boxplots After IQR Capping")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Target: cnt = total bike rentals in that hour -> regression problem
target_column = "cnt"
# Categorical columns represented by numeric category codes in this dataset
# (season was already one-hot encoded above; weathersit is encoded inside the pipeline later)
categorical_columns = [
    column for column in ["weathersit"]
    if column in df.columns
]
print("Categorical predictor columns:", categorical_columns)
```

```python
# casual and registered are excluded: together they add up to the target.
X = df.drop(
    columns=[target_column, "casual", "registered"],
    errors="ignore"
)
y = df[target_column].astype(float)
print("Independent variables (X):", X.shape)
print("Dependent variable (y):", y.shape)
print("\nTarget summary:")
print(y.describe().round(2))
```

```python
# Rename columns for easier reference.
rename_map = {
    "yr": "year", "mnth": "month", "hr": "hour",
    "hum": "humidity", "atemp": "feels_like_temp", "cnt": "total_rentals"
}
df = df.rename(columns=rename_map)
X = X.rename(columns=rename_map)
y.name = "total_rentals"
print("Updated column names:")
print(df.columns.tolist())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
  X,
  y,
  test_size=0.20,
  random_state=RANDOM_STATE
)
print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("\nTraining target mean / std:", round(y_train.mean(), 2), "/", round(y_train.std(), 2))
print("Testing target mean / std:", round(y_test.mean(), 2), "/", round(y_test.std(), 2))
```

```python
numeric_features = X_train.select_dtypes(include="number").columns.tolist()
categorical_features = [
   column for column in categorical_columns if column in X_train.columns
]
# Remove categorical columns from the numerical feature list.
numeric_features = [
   column for column in numeric_features
   if column not in categorical_features
]
numeric_transformer = Pipeline(steps=[
     ("imputer", SimpleImputer(strategy="median")),
     ("scaler", StandardScaler())
])
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])
preprocessor = ColumnTransformer(
   transformers=[
     ("numeric", numeric_transformer, numeric_features),
     ("categorical", categorical_transformer, categorical_features)
   ],
   remainder="drop"
)
print("Numerical features:", numeric_features)
print("Categorical features:", categorical_features)
```

```python
# Fit on the training set only, then apply the same steps to the test set
X_train_scaled = preprocessor.fit_transform(X_train)
X_test_scaled = preprocessor.transform(X_test)
# names of the columns after preprocessing (weathersit becomes weathersit_1, _2, _3, _4)
feature_names = [name.split("__")[-1] for name in preprocessor.get_feature_names_out()]
print("Train shape:", X_train_scaled.shape)
print("Test shape:", X_test_scaled.shape)
print("Features after preprocessing:", feature_names)
```

```python
# Feature selection: keep the 8 best features (fitted on the training set only)
selector = SelectKBest(score_func=f_regression, k=8)
X_new = selector.fit_transform(X_train_scaled, y_train)
```

```python
selected_features = np.array(feature_names)[selector.get_support()]
print("Selected Features:")
print(selected_features)
```

```python
feature_scores = pd.DataFrame({
  'Feature': feature_names,
  'Score': selector.scores_,
  'Selected': selector.get_support()
})
feature_scores = feature_scores.sort_values(
  by='Score',
  ascending=False
)
print(feature_scores)
```

```python
# Selecting Algorithm
models = {
   "Linear Regression": LinearRegression(),
   "Decision Tree": DecisionTreeRegressor(random_state=RANDOM_STATE),
   "Random Forest": RandomForestRegressor(n_estimators=100, random_state=RANDOM_STATE),
}
```

```python
#Store results
scores = []
for name, model in models.items():
   model.fit(X_train_scaled, y_train)
   y_pred = model.predict(X_test_scaled)
   mae = mean_absolute_error(y_test, y_pred)
   rmse = np.sqrt(mean_squared_error(y_test, y_pred))
   r2 = r2_score(y_test, y_pred)
   scores.append({
     "Model": name,
     "MAE": round(mae, 3),
     "RMSE": round(rmse, 3),
     "R2": round(r2, 3)
   })
# Convert to DataFrame for nice display
results_df = pd.DataFrame(scores)
print(results_df)
```

```python
results_df
```

```python
# Best model by R2: actual vs predicted
best_name = results_df.sort_values("R2", ascending=False).iloc[0]["Model"]
best_model = models[best_name]
y_pred_best = best_model.predict(X_test_scaled)
print("Best model:", best_name)
print("R2 :", round(r2_score(y_test, y_pred_best), 3))
print("MAE :", round(mean_absolute_error(y_test, y_pred_best), 2), "bikes per hour")
plt.figure(figsize=(6, 6))
plt.scatter(y_test, y_pred_best, alpha=0.3)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], color="red")
plt.title(f"Actual vs Predicted Rentals - {best_name}")
plt.xlabel("Actual rentals")
plt.ylabel("Predicted rentals")
plt.tight_layout()
plt.show()
```

```python
# Which features matter most to the Random Forest?
importances = pd.Series(
  models["Random Forest"].feature_importances_,
  index=feature_names
).sort_values()
plt.figure(figsize=(8, 6))
importances.plot(kind="barh")
plt.title("Random Forest Feature Importance")
plt.tight_layout()
plt.show()
```
